```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 기본 함수 f(x) 정의
def f(x):
    condlist = [
        x < 0,
        (0 <= x) & (x < 0.5),
        (0.5 <= x) & (x < 1),
        x >= 1
    ]
    choicelist = [
        0,
        2 * x,
        2 - 2 * x,
        0
    ]
    return np.select(condlist, choicelist)

def g(x):
    return f(x)

def g_double(x):
    return g(g(x))

def g_triple(x):
    return g(g(g(x)))

# 2. 미분 불가능한 점(꺾임점)의 위치를 자동으로 생성하는 함수
def get_non_diff_points(n):
    # n번 합성했을 때 꺾이는 지점들은 0부터 1까지 1/(2^n) 간격으로 존재합니다.
    num_segments = 2**n
    return [i / num_segments for i in range(num_segments + 1)]

# 각 함수의 미분 불가능한 점 좌표 자동 계산
non_diff_f = get_non_diff_points(1)   # 1번 (f(x)): [0.0, 0.5, 1.0]
non_diff_g2 = get_non_diff_points(2)  # 2번 합성 (g(g(x))): [0.0, 0.25, 0.5, 0.75, 1.0]
non_diff_g3 = get_non_diff_points(3)  # 3번 합성 (g(g(g(x)))): [0.0, 0.125, 0.25, ...]

# 3. 콘솔창에 미분 불가능한 점의 개수 출력
print("==================================================")
print("               미분 불가능한 점 개수 결과")
print("==================================================")
print(f"g(g(g(x)))의 미분 불가능한 점 개수: {len(non_diff_g3)}개")
print(f"   - 좌표: {non_diff_g3}")
print("==================================================")

# 4. 그래프 그리기를 위한 x값 설정
x_vals = np.linspace(-0.3, 1.3, 1000)

plt.figure(figsize=(12, 10))

# [첫 번째 그래프] f(x)
plt.subplot(3, 1, 1)
plt.plot(x_vals, f(x_vals), label='$f(x)$', color='blue', linewidth=2)
plt.scatter(non_diff_f, f(np.array(non_diff_f)), color='red', marker='o', s=80, edgecolor='black', zorder=5,
            label=f'Non-differentiable ({len(non_diff_f)} pts)')
plt.title('Original Function $f(x)$', fontsize=13)
plt.ylabel('$y$',rotation=360)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks([0, 0.5, 1.0])
plt.xlim(-0.2, 1.2)
plt.ylim(0, 1.3)
plt.legend(loc='upper right')

# [두 번째 그래프] g(g(x))
plt.subplot(3, 1, 2)
plt.plot(x_vals, g_double(x_vals), label='$g(g(x))$', color='orange', linewidth=2)
plt.scatter(non_diff_g2, g_double(np.array(non_diff_g2)), color='red', marker='o', s=80, edgecolor='black', zorder=5,
            label=f'Non-differentiable ({len(non_diff_g2)} pts)')
plt.title('Double Composition $g(g(x))$', fontsize=13)
plt.ylabel('$y$',rotation=360)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks([0, 0.25, 0.5, 0.75, 1.0])
plt.xlim(-0.2, 1.2)
plt.ylim(0, 1.3)
plt.legend(loc='upper right')

# [세 번째 그래프] g(g(g(x)))
plt.subplot(3, 1, 3)
plt.plot(x_vals, g_triple(x_vals), label='$g(g(g(x)))$', color='red', linewidth=2)
plt.scatter(non_diff_g3, g_triple(np.array(non_diff_g3)), color='blue', marker='o', s=80, edgecolor='black', zorder=5,
            label=f'Non-differentiable ({len(non_diff_g3)} pts)')
plt.title('Triple Composition $g(g(g(x)))$', fontsize=13)
plt.xlabel('$x$')
plt.ylabel('$y$',rotation=360)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks([0, 0.125, 0.25, 0.375, 0.5, 0.625, 0.75, 0.875, 1.0])
plt.xlim(-0.2, 1.2)
plt.ylim(0, 1.3)
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 기본 텐트 함수 f(x) 정의
def f(x):
    condlist = [
        x < 0,
        (0 <= x) & (x < 0.5),
        (0.5 <= x) & (x < 1),
        x >= 1
    ]
    choicelist = [
        0,
        2 * x,
        2 - 2 * x,
        0
    ]
    return np.select(condlist, choicelist)

def g(x, a):
    return a * f(x)

def g_triple(x, a):
    return g(g(g(x, a), a), a)

# 2. 진짜 미분 불가능한 점을 찾는 함수 (개선 버전)
def find_true_peaks(a_val):
    # a <= 0 이면 함수가 모든 구간에서 y=0 이 되므로 미분불가능점이 없습니다.
    if a_val <= 0:
        return np.array([]) # 빈 배열 반환 (0개)

    # 정밀한 탐색을 위해 촘촘한 x 배열 생성
    x = np.linspace(0, 1, 50000)
    y = g_triple(x, a_val)

    # 기울기 계산
    dy = np.diff(y)
    dx = np.diff(x)
    slope = dy / dx

    # 기울기 변화량 감지
    slope_diff = np.abs(np.diff(slope))

    threshold = 0.5
    peak_indices = np.where(slope_diff > threshold)[0] + 1

    true_x_indices = []
    if len(peak_indices) > 0:
        groups = np.split(peak_indices, np.where(np.diff(peak_indices) > 5)[0] + 1)
        for group in groups:
            true_x_indices.append(int(np.median(group)))

    # a > 0 일 때만 경계점 포함하여 정렬
    critical_x = [0.0]
    for idx in true_x_indices:
        critical_x.append(x[idx])
    critical_x.append(1.0)

    return np.unique(np.round(critical_x, 5))

# 3. 부호 및 크기 비교 함수
def compare_values(y, target):
    if np.isclose(y, target, atol=1e-4):
        return "="
    elif y > target:
        return ">"
    else:
        return "<"

# 4. 범위 설정
a_cases = {
    "1) a <= 0 (a = -0.5)": -0.5,
    "2) 0 < a <= 1/2 (a = 0.4)": 0.4,
    "3) 1/2 < a < 1 (a = 0.75)": 0.75,
    "4) a > 1 (a = 1.2)": 1.2
}

x_plot = np.linspace(-0.1, 1.1, 3000)
plt.figure(figsize=(15, 12))

print("==========================================================================")
print("             정밀 기울기 추적 기반 미분 불가능한 점 판정 결과 (수정본)")
print("==========================================================================")

for idx, (range_title, a) in enumerate(a_cases.items(), 1):
    crit_x = find_true_peaks(a)

    print(f"\n■ {range_title}")
    print(f"  * 탐지된 미분 불가능한 점 개수: {len(crit_x)}개")

    if len(crit_x) > 0:
        crit_y = g_triple(crit_x, a)
        for cx, cy in zip(crit_x, crit_y):
            sign_0 = compare_values(cy, 0)
            sign_a = compare_values(cy, a)
            print(f"    - 점 x = {cx:5.3f} -> 함숫값 y = {cy:5.3f}  [ y {sign_0} 0 ]  [ y {sign_a} a ]")
    else:
        print("    - 미분 불가능한 점이 존재하지 않는 매끄러운(평평한) 함수입니다.")

    # --- 그래프 시각화 ---
    plt.subplot(2, 2, idx)
    plt.plot(x_plot, g_triple(x_plot, a), color='teal', linewidth=2, label='$g(g(g(x)))$')

    if len(crit_x) > 0:
        plt.scatter(crit_x, g_triple(crit_x, a), color='gold', marker='o', s=80, edgecolor='black', linewidth=1.5, zorder=5,
                    label=f'True Peaks ({len(crit_x)} pts)')

    plt.title(range_title, fontsize=13, fontweight='bold')
    plt.xlabel('$x$')
    plt.ylabel('$y$')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.axhline(0, color='black', linewidth=0.8)

    if a <= 0: plt.ylim(-0.3, 0.5)
    elif a > 1: plt.ylim(-0.1, a + 0.3)
    else: plt.ylim(-0.1, 1.1)
    plt.xlim(-0.05, 1.05)
    plt.legend(loc='upper right')

print("\n==========================================================================")
plt.suptitle('Corrected Non-differentiable Points Analysis of $g(g(g(x)))$', fontsize=16, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.integrate import simpson

# 1. 기본 텐트 함수 f(x) 정의
def f(x):
    condlist = [
        x < 0,
        (0 <= x) & (x < 0.5),
        (0.5 <= x) & (x < 1),
        x >= 1
    ]
    choicelist = [0, 2 * x, 2 - 2 * x, 0]
    return np.select(condlist, choicelist)

def g(x, a):
    return a * f(x)

def g_triple(x, a):
    return g(g(g(x, a), a), a)

# 2. 정밀한 풀이 과정 콘솔 출력
def print_exact_proof():
    proof = """
================================================================================
 [수학적 풀이 과정] 기하학적 도형 특징을 이용한 g(g(g(x)))와 g(x)의 넓이 매칭
================================================================================

방정식: Integral_0^1 g(g(g(x))) dx = Integral_0^1 g(x) dx

[우변 분석] g(x)의 넓이 (S_right)
- g(x) = a * f(x) 이며, [0, 1] 구간에서 밑변이 1이고 높이가 a인 삼각형입니다.
- S_right = 1/2 * 밑변 * 높이 = 1/2 * 1 * a = 0.5 * a

[좌변 분석 및 해(a) 도출] g(g(g(x)))의 형태에 따른 조건 만족 조건

1. a = 0 일 때 (상수함수 화)
   - g(x) = 0 이 되므로 세 번 합성해도 모든 구간에서 함수는 y = 0 입니다.
   - S_left = 0, S_right = 0 으로 일치합니다.
   => 해 1: [ a = 0 ]

2. a = 1/2 일 때 (기하학적 축소 평형)
   - g(x)의 최댓값이 1/2이 되며, 세 번 합성한 g(g(g(x))) 또한 높이가 1/2로 제한됩니다.
   - 그래프는 [0, 1] 구간 전체를 밑변으로 하고 높이가 1/2인 대칭 삼각형 1개를 유지합니다.
   - S_left  = 1/2 * 1 * 1/2 = 1/4 (0.25)
   - S_right = 1/2 * (1/2)   = 1/4 (0.25)
   - 두 넓이가 0.25로 완벽히 일치하므로 성립합니다.
   => 해 2: [ a = 1/2 ]

3. a = 1 일 때 (완벽한 프랙탈 주기 복제)
   - g(x)의 최댓값이 1이 되어, 함숫값이 정역을 100% 활용합니다.
   - 세 번 합성하면 높이가 1로 유지된 채 밑변이 1/4인 똑같은 삼각형 4개가 채워집니다.
   - S_left  = 4 * (1/2 * 1/4 * 1) = 1/2 (0.5)
   - S_right = 1/2 * 1             = 1/2 (0.5)
   - 두 넓이가 0.5로 완벽히 일치하므로 성립합니다.
   => 해 3: [ a = 1 ]

결론적으로, 구하고자 하는 특수 경계 조건의 해는 다음과 같습니다:
★ 정답: a = 0, a = 1/2 (0.5), a = 1
================================================================================
    """
    print(proof)

# 3. 메인 실행부 및 수치 검증
if __name__ == "__main__":
    print_exact_proof()

    # 0부터 1까지 촘촘한 정적분 데이터 생성
    x_vals = np.linspace(0, 1, 10001)
    target_a_values = [0.0, 0.5, 1.0]

    print("   [파이썬 코드를 통한 정밀 수치적분(Simpson's Rule) 검증]")
    print(f"   {'지정된 a 값':^12} | {'g(x) 실제 넓이':^14} | {'g(g(g(x))) 실제 넓이':^18} | {'결과':^6}")
    print("   " + "-" * 65)

    for a in target_a_values:
        area_g = simpson(g(x_vals, a), x_vals)
        area_g3 = simpson(g_triple(x_vals, a), x_vals)

        # 소수점 오차 감안하여 비교
        is_match = "성립" if np.isclose(area_g, area_g3, atol=1e-4) else "불일치"
        print(f"   {a:12.1f} | {area_g:14.4f} | {area_g3:18.4f} | {is_match:^6}")
    print("   =================================================================")
```

```python
# 1. 기본 텐트 함수 f(x) 및 g(x) 계열 정의 (수치적 구조 확인용)
import numpy as np

def f(x):
    condlist = [x < 0, (0 <= x) & (x < 0.5), (0.5 <= x) & (x < 1), x >= 1]
    choicelist = [0, 2 * x, 2 - 2 * x, 0]
    return np.select(condlist, choicelist)

def g(x, a):
    return a * f(x)

def g_triple(x, a):
    return g(g(g(x, a), a), a)


# 2. 전 구간 미존재 증명 과정 콘솔 출력 함수
def print_non_existence_proof():
    proof = """
================================================================================
 [수학적 증명] 경계점을 제외한 4가지 구간에서 해(a)가 존재하지 않는 이유
================================================================================

우리가 비교할 두 도형의 넓이 식은 다음과 같습니다:
- 우변 S_right (g(x)의 넓이)           = 0.5 * a
- 좌변 S_left  (g(g(g(x)))의 전체 넓이) = (a의 값에 따라 파형과 넓이가 결정됨)

각 구간별로 두 넓이가 절대 같아질 수 없음을 기하학적으로 증명합니다.

--------------------------------------------------------------------------------
구간 ①: a < 0 (음수 영역)
--------------------------------------------------------------------------------
- [기하학적 현상]: g(x)의 치역이 음수가 되므로, 한 번만 합성해도 f(x)의 x < 0 정의에 의해
  모든 함숫값이 y = 0 인 상수함수로 완전히 평평해집니다.
- [넓이 비교]: S_left = 0  vs  S_right = 0.5 * a (음수)
- [결론]: 0 = 0.5 * a 를 만족하려면 a = 0 이어야 하므로, a < 0 영역에는 해가 존재하지 않습니다.

--------------------------------------------------------------------------------
구간 ②: 0 < a < 1/2 (낮은 단일 산 영역)
--------------------------------------------------------------------------------
- [기하학적 현상]: g(x)의 최댓값(높이)이 0.5 미만입니다. 이 값이 다음 합성의 입력값으로
  들어갈 때, f(x)의 정점(0.5)을 넘지 못하므로 산이 여러 개로 복제되지 못하고 중심에 1개만 유지됩니다.
- [높이와 넓이]: 매 합성마다 높이가 2a배씩 수축하므로, 최종 g(g(g(x)))의 높이는 2 * a^3 이 됩니다.
  도형은 밑변이 1이고 높이가 2*a^3 인 삼각형이므로 넓이는 S_left = a^3 입니다.
- [넓이 비교]: a^3 = 0.5 * a  =>  a^2 = 0.5  =>  a = 1/sqrt(2) ≒ 0.707
- [결론]: 계산된 해(0.707)는 현재 구간인 0 < a < 0.5 를 완전히 벗어나므로, 이 구간 내에서는
  만족하는 해가 존재하지 않습니다.

--------------------------------------------------------------------------------
구간 ③: 1/2 < a < 1 (프랙탈 톱니 영역의 특징)
--------------------------------------------------------------------------------
- [주의]: 이 구간 내부에서는 함수가 완벽한 프랙탈 대칭을 이루어 S_left = 0.5 * a 로
  우변과 넓이가 항상 같은 '항등식 상태'가 됩니다.
- [결론]: 따라서 질문하신 개구간(Open Interval) 1/2 < a < 1 '자체'는 수식적 균형이 성립하는
  연속적인 해의 구간입니다. (특정 단일 원소 해로 분리되지 않고 구간 전체가 참입니다.)

--------------------------------------------------------------------------------
구간 ④: a > 1 (상단 절단 및 붕괴 영역)
--------------------------------------------------------------------------------
- [기하학적 현상]: g(x)의 최댓값이 1을 초과해 버립니다. f(x)의 정의상 입력값이 1을 넘으면
  출력값이 무조건 0으로 가라앉기 때문에, 톱니 모양의 윗부분이 잘려 나가며 바닥(y=0)으로
  급격히 떨어지는 구멍(슬릿)들이 생성됩니다.
- [넓이 비교]: 높이가 유실되고 파형이 잘려 나가면서 좌변의 넓이가 급격하게 손실됩니다.
  그 결과 항상 [ S_left < 0.5 * a ] 상태가 됩니다.
- [결론]: 좌변이 우변보다 무조건 작아지므로, a > 1 영역에서는 해가 절대로 존재할 수 없습니다.

================================================================================
 요약: 단일 값으로 딱 떨어지는 경계 조건(a=0, 1/2, 1) 외에,
       나머지 개구간 영역 ①, ②, ④ 에서는 방정식이 성립하는 a가 단 하나도 존재하지 않습니다.
================================================================================
"""
    print(proof)

# 3. 코드 실행
if __name__ == "__main__":
    print_non_existence_proof()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson

# 1. Define Tent Function f(x) and Parameterized Function g(x)
def f(x):
    condlist = [
        x < 0,
        (0 <= x) & (x < 0.5),
        (0.5 <= x) & (x < 1),
        x >= 1
    ]
    choicelist = [0, 2 * x, 2 - 2 * x, 0]
    return np.select(condlist, choicelist)

def g(x, a):
    return a * f(x)

def g_triple(x, a):
    return g(g(g(x, a), a), a)

# 2. X-axis Arrays for Plotting and Integration
x_plot = np.linspace(-0.1, 1.1, 2000)
x_integral = np.linspace(0, 1, 1000)  # Interval [0, 1]

# Solution cases to analyze
a_solutions = {
    "1) a = 0.0 (Trivial Solution)": 0.0,
    "2) a = 0.5 (Geometric Equilibrium Solution)": 0.5,
    "3) a = 1.0 (Perfect Fractal Replication Solution)": 1.0
}

# 3. Create Subplots (3 Rows, 2 Columns)
plt.figure(figsize=(15, 14))

for idx, (title, a) in enumerate(a_solutions.items(), 1):
    # Calculate exact area using Simpson's rule
    int_g = simpson(g(x_integral, a), x_integral)
    int_g3 = simpson(g_triple(x_integral, a), x_integral)

    # --- [Left Column] Area under g(x) ---
    plt.subplot(3, 2, idx*2 - 1)
    plt.plot(x_plot, g(x_plot, a), color='royalblue', linewidth=2, label='$g(x)$')

    if a == 0.0:
        # Warning box for a = 0.0
        plt.text(0.5, 0.2, "When a = 0, y = 0 for all intervals.\nThe definite integral area cannot be visualized.",
                 ha='center', va='center', color='darkred', fontsize=11, fontweight='bold',
                 bbox=dict(boxstyle='round,pad=0.5', facecolor='whitesmoke', edgecolor='red', alpha=0.8))
    else:
        plt.fill_between(x_integral, g(x_integral, a), 0, color='royalblue', alpha=0.25,
                         label=f'Area = {int_g:.4f}')

    plt.title(f"{title}\n$g(x)$ Area Visualization", fontsize=12, fontweight='bold')
    plt.xlabel('$x$')
    plt.ylabel('$y$')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.axhline(0, color='black', linewidth=0.8)
    plt.xlim(-0.05, 1.05)
    plt.ylim(-0.1, max(a, 0.5) + 0.2)
    plt.legend(loc='upper right')

    # --- [Right Column] Area under g(g(g(x))) ---
    plt.subplot(3, 2, idx*2)
    plt.plot(x_plot, g_triple(x_plot, a), color='forestgreen', linewidth=2, label='$g(g(g(x)))$')

    if a == 0.0:
        # Warning box for a = 0.0
        plt.text(0.5, 0.2, "When a = 0, y = 0 for all intervals.\nThe definite integral area cannot be visualized.",
                 ha='center', va='center', color='darkred', fontsize=11, fontweight='bold',
                 bbox=dict(boxstyle='round,pad=0.5', facecolor='whitesmoke', edgecolor='red', alpha=0.8))
    else:
        plt.fill_between(x_integral, g_triple(x_integral, a), 0, color='forestgreen', alpha=0.25,
                         label=f'Area = {int_g3:.4f}')

    plt.title(f"{title}\n$g(g(g(x)))$ Area Visualization", fontsize=12, fontweight='bold')
    plt.xlabel('$x$')
    plt.ylabel('$y$')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.axhline(0, color='black', linewidth=0.8)
    plt.xlim(-0.05, 1.05)
    plt.ylim(-0.1, max(a, 0.5) + 0.2)
    plt.legend(loc='upper right')

plt.suptitle('Definite Integral Areas at Complete Solution Points', fontsize=16, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```
